````python
# Write your code to make your dataset analysis ready.

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

numeric_columns = df.select_dtypes(include=np.number).columns

print(numeric_columns)

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()

print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
display(df.dtypes)
````

````python

# Set visual style
sns.set_theme(style='whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

# 1. Clean meta_df (reviews) - Corrected to use meta_df for review-related cleaning
meta_df['Rating_clean'] = pd.to_numeric(meta_df['Rating'], errors='coerce')
meta_df['Time_dt'] = pd.to_datetime(meta_df['Time'], errors='coerce')
meta_df['Year'] = meta_df['Time_dt'].dt.year
meta_df['Month_Year'] = meta_df['Time_dt'].dt.to_period('M')
meta_df['DayOfWeek'] = meta_df['Time_dt'].dt.day_name()
meta_df['Hour'] = meta_df['Time_dt'].dt.hour
meta_df['Review_Length'] = (
    meta_df['Review']
    .astype(str)
    .apply(lambda x: len(x.split()) if pd.notna(x) else 0)
)

# Extract reviewer metrics from Metadata string
def parse_metadata(meta_str):
  if pd.isna(meta_str):
    return np.nan, np.nan
  parts = str(meta_str).split(',')
  revs, folls = np.nan, np.nan
  for p in parts:
    p = p.strip()
    if 'Review' in p:
      try:
        revs = float(p.split()[0])
      except:
        pass
    elif 'Follower' in p:
      try:
        folls = float(p.split()[0])
      except:
        pass
  return revs, folls

meta_df[['Reviewer_Reviews', 'Reviewer_Followers']] = meta_df['Metadata'].apply(
    lambda x: pd.Series(parse_metadata(x))
)

# 2. Clean df (metadata) - Corrected to use df for metadata-related cleaning
df['Cost_clean'] = pd.to_numeric(
    df['Cost'].astype(str).str.replace(',', ''), errors='coerce'
)
df['Num_Cuisines'] = (
    df['Cuisines'].dropna().apply(lambda x: len(x.split(',')))
)
df['Has_Collection'] = (
    df['Collections'].notna().map({True: 'Featured', False: 'Regular'})
)

# 3. Restaurant-level aggregation & merge
rest_summary = (
    meta_df.groupby('Restaurant') # Grouping by 'Restaurant' from meta_df
    .agg(
        Avg_Rating=('Rating_clean', 'mean'),
        Total_Reviews=('Rating_clean', 'count'),
        Total_Pictures=('Pictures', 'sum'),
        Avg_Review_Length=('Review_Length', 'mean'),
    )
    .reset_index()
)

merged_rest = pd.merge(
    df, rest_summary, left_on='Name', right_on='Restaurant', how='inner' # Merging df (metadata) with aggregated review summary
)

plt.figure(figsize=(9, 5))
sns.histplot(data=meta_df, x='Rating_clean', bins=10, kde=True, color='#2b5c8f') # Plotting from meta_df
plt.title('Distribution of Customer Ratings', fontsize=13, weight='bold', pad=12)
plt.xlabel('Rating (out of 5)', fontsize=11)
plt.ylabel('Number of Reviews', fontsize=11)
plt.tight_layout()
plt.show()
````

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis ($H_0$): There is no significant difference in the average rating between restaurants featured in Zomato curated Collections and restaurants not featured in Collections ($\mu_{\text{featured}} = \mu_{\text{non-featured}}$).Alternate Hypothesis ($H_1$): Restaurants featured in Zomato curated Collections have a significantly higher average rating than restaurants not featured in Collections ($\mu_{\text{featured}} > \mu_{\text{non-featured}}$).Significance Level ($\alpha$): 0.05

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy import stats

# 1. Prepare groups
featured_ratings = merged_rest[merged_rest['Has_Collection'] == 1][
    'Avg_Rating'
].dropna()
non_featured_ratings = merged_rest[merged_rest['Has_Collection'] == 0][
    'Avg_Rating'
].dropna()

# 2. Perform Two-Sample One-Tailed Welch's t-test
t_stat, p_val = stats.ttest_ind(
    featured_ratings,
    non_featured_ratings,
    equal_var=False,
    alternative='greater',
)

# 3. Print Results
print('--- Hypothesis Test 1 Results ---')
print(
    f'Featured Group Mean: {featured_ratings.mean():.4f} (Std:'
    f' {featured_ratings.std():.4f}, N: {len(featured_ratings)})'
)
print(
    f'Non-Featured Group Mean: {non_featured_ratings.mean():.4f} (Std:'
    f' {non_featured_ratings.std():.4f}, N: {len(non_featured_ratings)})'
)
print(f't-statistic: {t_stat:.4f}')
print(f'p-value: {p_val:.6f}')

if p_val < 0.05:
  print('Decision: Reject the Null Hypothesis (H0).')
else:
  print('Decision: Fail to Reject the Null Hypothesis (H0).')
````

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments

# Visualizing & capping numerical features using IQR Winsorization
outlier_cols = ['Cost_clean', 'Pictures']


def cap_outliers_iqr(data, columns):
  df_capped = data.copy()
  for col in columns:
    q1 = df_capped[col].quantile(0.25)
    q3 = df_capped[col].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    # Cap values at lower and upper thresholds
    df_capped[col] = np.clip(df_capped[col], lower_limit, upper_limit)
  return df_capped


full_df = cap_outliers_iqr(full_df, outlier_cols)
print(
    'Summary statistics after outlier capping:\n',
    full_df[outlier_cols].describe(),
)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Interquartile Range (IQR) Winsorization / Capping: Instead of trimming (deleting) rows containing luxury dining costs or viral photo counts—which would discard valid high-end restaurant data—values beyond $[Q_1 - 1.5 \times \text{IQR}, Q_3 + 1.5 \times \text{IQR}]$ were capped.Robustness: Capping bounds the influence of extreme values on linear model weights and distance metrics while retaining complete dataset rows.

````python
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report
)
import pandas as pd
import numpy as np
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.utils import resample

# 1. Load data (ensure these CSVs are available in the environment)
reviews_df = pd.read_csv('/Zomato Restaurant reviews.csv')
meta_df_restaurants = pd.read_csv('/Zomato Restaurant names and Metadata.csv') # Renamed to avoid conflict

# Clean base (using reviews_df for review data, meta_df_restaurants for restaurant metadata)
df = reviews_df.copy()
df['Rating_clean'] = pd.to_numeric(df['Rating'], errors='coerce')
df = df.dropna(subset=['Rating_clean', 'Review']).reset_index(drop=True)

meta_df_restaurants['Cost_clean'] = pd.to_numeric(meta_df_restaurants['Cost'].astype(str).str.replace(',', ''), errors='coerce')
meta_df_restaurants['Cost_clean'] = meta_df_restaurants['Cost_clean'].fillna(meta_df_restaurants['Cost_clean'].median())

full_df = pd.merge(df, meta_df_restaurants, left_on='Restaurant', right_on='Name', how='left')

# Extract metadata stats
def parse_meta(meta_str):
    if pd.isna(meta_str):
        return 0, 0
    parts = str(meta_str).split(',')
    revs, folls = 0, 0
    for p in parts:
        p = p.strip()
        if 'Review' in p:
            try: revs = float(p.split()[0])
            except: pass
        elif 'Follower' in p:
            try: folls = float(p.split()[0])
            except: pass
    return revs, folls

full_df[['Reviewer_Reviews', 'Reviewer_Followers']] = full_df['Metadata'].apply(lambda x: pd.Series(parse_meta(x)))

# Features
full_df['Review_Word_Count'] = full_df['Review'].astype(str).apply(lambda x: len(x.split()))
full_df['Review_Char_Count'] = full_df['Review'].astype(str).apply(len)
full_df['Avg_Word_Length'] = full_df['Review_Char_Count'] / (full_df['Review_Word_Count'] + 1e-5)
full_df['Time_dt'] = pd.to_datetime(full_df['Time'], errors='coerce')
full_df['Is_Weekend'] = full_df['Time_dt'].dt.day_name().isin(['Saturday', 'Sunday']).astype(int)
full_df['Has_Collection'] = full_df['Collections'].notna().astype(int)
full_df['Num_Cuisines'] = full_df['Cuisines'].fillna('').apply(lambda x: len(x.split(',')) if x else 0)
full_df['Pictures'] = full_df['Pictures'].fillna(0)
full_df['Sentiment_Target'] = (full_df['Rating_clean'] >= 4).astype(int)

# Log transform
for col in ['Cost_clean', 'Review_Word_Count', 'Reviewer_Reviews', 'Reviewer_Followers']:
    full_df[f'Log_{col}'] = np.log1p(full_df[col])

dense_cols = ['Log_Cost_clean', 'Log_Review_Word_Count', 'Avg_Word_Length', 'Log_Reviewer_Reviews', 'Log_Reviewer_Followers', 'Pictures', 'Has_Collection', 'Num_Cuisines', 'Is_Weekend']
scaler = StandardScaler()
X_dense = scaler.fit_transform(full_df[dense_cols])

# TF-IDF & TruncatedSVD
tfidf = TfidfVectorizer(max_features=2500, stop_words='english', ngram_range=(1, 2), min_df=3)
X_tfidf = tfidf.fit_transform(full_df['Review'].fillna(''))
svd = TruncatedSVD(n_components=50, random_state=42)
X_svd = svd.fit_transform(X_tfidf)

X_all = np.hstack((X_dense, X_svd))
y_all = full_df['Sentiment_Target'].values

# Split
X_train, X_test, y_train, y_test = train_test_split(X_all, y_all, test_size=0.20, random_state=42, stratify=y_all)

# Train balancing
train_data = pd.DataFrame(X_train)
train_data['target'] = y_train
majority = train_data[train_data['target'] == 1]
minority = train_data[train_data['target'] == 0]
minority_upsampled = resample(minority, replace=True, n_samples=len(majority), random_state=42)
balanced_train = pd.concat([majority, minority_upsampled])
X_train_bal = balanced_train.drop(columns='target').values
y_train_bal = balanced_train['target'].values

# Initialize the Algorithm
xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42,
    eval_metric='logloss',
    n_jobs=-1
)

# Fit the Algorithm on balanced training data
xgb_model.fit(X_train_bal, y_train_bal)

# Predict on the model (Test set)
y_test_pred_xgb = xgb_model.predict(X_test)
y_test_prob_xgb = xgb_model.predict_proba(X_test)[:, 1]

# Calculate Baseline Metrics
xgb_baseline_metrics = {
    'Accuracy': accuracy_score(y_test, y_test_pred_xgb),
    'Precision': precision_score(y_test, y_test_pred_xgb),
    'Recall': recall_score(y_test, y_test_pred_xgb),
    'F1-Score': f1_score(y_test, y_test_pred_xgb),
    'ROC-AUC': roc_auc_score(y_test, y_test_prob_xgb)
}

print("--- ML Model 3 (XGBoost) Classification Report ---")
print(
    classification_report(
        y_test,
        y_test_pred_xgb,
        target_names=['Negative/Neutral (<4)', 'Positive (>=4)']
    )
)

print("\n--- XGBoost Baseline Metrics ---")
for metric, value in xgb_baseline_metrics.items():
    print(f"{metric}: {value:.4f}")
````

````python
# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model

from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

# Initialize XGBoost
xgb_model = XGBClassifier(
    random_state=42,
    eval_metric='logloss',
    n_jobs=-1
)

# Define hyperparameter grid
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# GridSearchCV
grid_search = GridSearchCV(
    estimator=xgb_model,
    param_grid=param_grid,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

# Fit the Algorithm
grid_search.fit(X_train_bal, y_train_bal)

# Display best parameters
print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(grid_search.best_score_)

# Best model
best_xgb_model = grid_search.best_estimator_

# Predict on the Test Set
y_test_pred_xgb_tuned = best_xgb_model.predict(X_test)
y_test_prob_xgb_tuned = best_xgb_model.predict_proba(X_test)[:, 1]

# Calculate Tuned Model Metrics
xgb_tuned_metrics = {
    'Accuracy': accuracy_score(y_test, y_test_pred_xgb_tuned),
    'Precision': precision_score(y_test, y_test_pred_xgb_tuned),
    'Recall': recall_score(y_test, y_test_pred_xgb_tuned),
    'F1-Score': f1_score(y_test, y_test_pred_xgb_tuned),
    'ROC-AUC': roc_auc_score(y_test, y_test_prob_xgb_tuned)
}

print("\n--- ML Model 3 (Tuned XGBoost) Classification Report ---")
print(
    classification_report(
        y_test,
        y_test_pred_xgb_tuned,
        target_names=['Negative/Neutral (<4)', 'Positive (>=4)']
    )
)

print("\n--- Tuned XGBoost Metrics ---")
for metric, value in xgb_tuned_metrics.items():
    print(f"{metric}: {value:.4f}")
````
