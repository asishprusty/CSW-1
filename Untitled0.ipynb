```python
from datetime import datetime
from langchain.tools import tool

@tool
def search_flights(source: str, destination: str) -> str:
    """Search for flights between a source city and a destination city, and return the cheapest and fastest options."""
    # Filter flights that match both the starting city and destination city
    matching_flights = [
        f for f in flights
        if f["from"].lower() == source.lower() and f["to"].lower() == destination.lower()
    ]

    if not matching_flights:
        return f"No flights found from {source} to {destination}."

    # Pick the flight with the lowest price
    cheapest = min(matching_flights, key=lambda f: f["price"])

    # Calculate how many hours each flight takes
    def get_duration_hours(flight):
        dep = datetime.fromisoformat(flight["departure_time"])
        arr = datetime.fromisoformat(flight["arrival_time"])
        return (arr - dep).total_seconds() / 3600

    # Pick the flight with the shortest duration
    fastest = min(matching_flights, key=get_duration_hours)

    return (
        f"Found {len(matching_flights)} flights from {source} to {destination}.\n"
        f"Cheapest: {cheapest['airline']} ({cheapest['flight_id']}) - ₹{cheapest['price']} "
        f"(Departs {cheapest['departure_time']}, Duration: {get_duration_hours(cheapest):.1f} hrs)\n"
        f"Fastest: {fastest['airline']} ({fastest['flight_id']}) - ₹{fastest['price']} "
        f"(Departs {fastest['departure_time']}, Duration: {get_duration_hours(fastest):.1f} hrs)"
    )

# Test our brand-new tool with a sample trip from Bangalore to Goa
print(search_flights.invoke({"source": "Bangalore", "destination": "Goa"}))
```

```python
@tool
def recommend_hotels(city: str, min_stars: int = 1, max_price: int = 10000) -> str:
    """Find hotels in a given city filtered by minimum star rating and maximum price per night."""
    # Keep only hotels that match the city, star rating, and budget
    matching_hotels = [
        h for h in hotels
        if h["city"].lower() == city.lower()
        and h["stars"] >= min_stars
        and h["price_per_night"] <= max_price
    ]

    if not matching_hotels:
        return f"No hotels found in {city} with at least {min_stars} stars under ₹{max_price}/night."

    # Sort the matches: highest stars first, then cheapest price first
    matching_hotels.sort(key=lambda h: (-h["stars"], h["price_per_night"]))

    # Turn the matching hotels into a neat, readable list
    result_lines = [f"Found {len(matching_hotels)} hotels in {city}:"]
    for h in matching_hotels:
        amenities = ", ".join(h["amenities"])
        result_lines.append(
            f"- {h['name']} ({h['hotel_id']}): {h['stars']}-star, ₹{h['price_per_night']}/night | Amenities: {amenities}"
        )

    return "\n".join(result_lines)

# Test our hotel tool in Goa (at least 4 stars, under ₹3000/night)
print(recommend_hotels.invoke({"city": "Goa", "min_stars": 4, "max_price": 3000}))
```

```python
@tool
def discover_places(city: str, place_type: str = "", min_rating: float = 0.0) -> str:
    """Find tourist attractions in a city, optionally filtered by type (e.g., beach, fort, temple, museum, park, lake, market) and minimum rating."""
    # Keep only places in the chosen city that meet the minimum rating
    matching_places = [
        p for p in places
        if p["city"].lower() == city.lower()
        and p["rating"] >= min_rating
    ]

    # If a specific type (like "fort") was requested, narrow the list down further
    if place_type:
        matching_places = [
            p for p in matching_places
            if p["type"].lower() == place_type.lower()
        ]

    if not matching_places:
        return f"No places found in {city} matching your criteria."

    # Sort so the highest-rated attractions appear first
    matching_places.sort(key=lambda p: p["rating"], reverse=True)

    # Turn the matches into a neat, readable list
    result_lines = [f"Found {len(matching_places)} places in {city}:"]
    for p in matching_places:
        result_lines.append(
            f"- {p['name']} ({p['place_id']}): Type: {p['type']}, Rating: {p['rating']}/5"
        )

    return "\n".join(result_lines)

# Test our places tool in Goa for attractions rated 4.1 or higher
print(discover_places.invoke({"city": "Goa", "min_rating": 4.1}))
```

```python
# A mini "phonebook" (dictionary) mapping our 8 cities to their GPS coordinates
CITY_COORDINATES = {
    "delhi": (28.6139, 77.2090),
    "mumbai": (19.0760, 72.8777),
    "goa": (15.2993, 74.1240),
    "bangalore": (12.9716, 77.5946),
    "chennai": (13.0827, 80.2707),
    "hyderabad": (17.3850, 78.4867),
    "kolkata": (22.5726, 88.3639),
    "jaipur": (26.9124, 75.7873),
}

def describe_weather_code(code: int) -> str:
    """Translate Open-Meteo's numeric weather codes into plain English."""
    if code == 0:
        return "Clear / Sunny"
    elif code in (1, 2, 3):
        return "Partly Cloudy"
    elif code in (45, 48):
        return "Foggy"
    elif code in (51, 53, 55, 61, 63, 65):
        return "Rainy"
    else:
        return "Showers / Unsettled"

@tool
def get_weather(city: str, days: int = 3) -> str:
    """Get the upcoming daily weather forecast for a city for a given number of days (1 to 7)."""
    coords = CITY_COORDINATES.get(city.lower())
    if not coords:
        return f"Sorry, I don't have coordinates for {city}."

    lat, lon = coords
    days = max(1, min(days, 7))  # Keep the number of days safely between 1 and 7

    url = (
        f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={lat}&longitude={lon}"
        f"&daily=temperature_2m_max,weathercode&timezone=auto"
    )

    # Safety net: catch any internet or API errors without crashing
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()["daily"]

        lines = [f"Weather forecast for {city.title()} (next {days} days):"]
        for i in range(days):
            date = data["time"][i]
            temp = data["temperature_2m_max"][i]
            condition = describe_weather_code(data["weathercode"][i])
            lines.append(f"- Day {i + 1} ({date}): {condition}, Max {temp}°C")

        return "\n".join(lines)
    except Exception as e:
        return f"Could not fetch weather for {city}: {e}"

# Test our weather tool for a 3-day trip to Goa
print(get_weather.invoke({"city": "Goa", "days": 3}))
```

```python
@tool
def estimate_budget(
    flight_price: int,
    hotel_price_per_night: int,
    nights: int,
    days: int,
    daily_local_expense: int = 1000
) -> str:
    """Calculate the total estimated trip budget by summing flight cost, hotel stay, and daily food/local travel expenses."""
    hotel_total = hotel_price_per_night * nights
    local_total = daily_local_expense * days
    total_cost = flight_price + hotel_total + local_total

    return (
        f"Estimated Budget Breakdown:\n"
        f"- Flight: ₹{flight_price:,}\n"
        f"- Hotel ({nights} nights @ ₹{hotel_price_per_night:,}/night): ₹{hotel_total:,}\n"
        f"- Food & Local Travel ({days} days @ ₹{daily_local_expense:,}/day): ₹{local_total:,}\n"
        f"- Total Estimated Cost: ₹{total_cost:,}"
    )

# Test our calculator using the cheap Goa flight (₹5356) and 5-star hotel (₹1232/night) we found earlier
print(estimate_budget.invoke({
    "flight_price": 5356,
    "hotel_price_per_night": 1232,
    "nights": 2,
    "days": 3,
    "daily_local_expense": 1000
}))
```

```python
%%writefile app.py
import json
import os
from datetime import datetime
import requests
import streamlit as st
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI

with open("flights.json", "r") as f:
    flights = json.load(f)

with open("hotels.json", "r") as f:
    hotels = json.load(f)

with open("places.json", "r") as f:
    places = json.load(f)

CITY_COORDINATES = {
    "delhi": (28.6139, 77.2090),
    "mumbai": (19.0760, 72.8777),
    "goa": (15.2993, 74.1240),
    "bangalore": (12.9716, 77.5946),
    "chennai": (13.0827, 80.2707),
    "hyderabad": (17.3850, 78.4867),
    "kolkata": (22.5726, 88.3639),
    "jaipur": (26.9124, 75.7873),
}

def describe_weather_code(code: int) -> str:
    """Convert Open-Meteo weather codes into readable labels."""
    if code == 0:
        return "Clear / Sunny"
    elif code in (1, 2, 3):
        return "Partly Cloudy"
    elif code in (45, 48):
        return "Foggy"
    elif code in (51, 53, 55, 61, 63, 65):
        return "Rainy"
    return "Showers / Unsettled"

@tool
def search_flights(source: str, destination: str) -> str:
    """Search for flights between a source city and a destination city, and return the cheapest and fastest options."""
    matching = [
        f for f in flights
        if f["from"].lower() == source.lower() and f["to"].lower() == destination.lower()
    ]
    if not matching:
        return f"No flights found from {source} to {destination}."

    cheapest = min(matching, key=lambda f: f["price"])

    def duration_hrs(flight):
        dep = datetime.fromisoformat(flight["departure_time"])
        arr = datetime.fromisoformat(flight["arrival_time"])
        return (arr - dep).total_seconds() / 3600

    fastest = min(matching, key=duration_hrs)
    return (
        f"Found {len(matching)} flights from {source} to {destination}.\n"
        f"Cheapest: {cheapest['airline']} ({cheapest['flight_id']}) - ₹{cheapest['price']} "
        f"(Departs {cheapest['departure_time']}, Duration: {duration_hrs(cheapest):.1f} hrs)\n"
        f"Fastest: {fastest['airline']} ({fastest['flight_id']}) - ₹{fastest['price']} "
        f"(Departs {fastest['departure_time']}, Duration: {duration_hrs(fastest):.1f} hrs)"
    )

@tool
def recommend_hotels(city: str, min_stars: int = 1, max_price: int = 10000) -> str:
    """Find hotels in a given city filtered by minimum star rating and maximum price per night."""
    matching = [
        h for h in hotels
        if h["city"].lower() == city.lower()
        and h["stars"] >= min_stars
        and h["price_per_night"] <= max_price
    ]
    if not matching:
        return f"No hotels found in {city} with at least {min_stars} stars under ₹{max_price}/night."

    matching.sort(key=lambda h: (-h["stars"], h["price_per_night"]))
    lines = [f"Found {len(matching)} hotels in {city}:"]
    for h in matching:
        lines.append(
            f"- {h['name']} ({h['hotel_id']}): {h['stars']}-star, ₹{h['price_per_night']}/night | Amenities: {', '.join(h['amenities'])}"
        )
    return "\n".join(lines)

@tool
def discover_places(city: str, place_type: str = "", min_rating: float = 0.0) -> str:
    """Find tourist attractions in a city, optionally filtered by type and minimum rating."""
    matching = [
        p for p in places
        if p["city"].lower() == city.lower() and p["rating"] >= min_rating
    ]
    if place_type:
        matching = [p for p in matching if p["type"].lower() == place_type.lower()]
    if not matching:
        return f"No places found in {city} matching your criteria."

    matching.sort(key=lambda p: p["rating"], reverse=True)
    lines = [f"Found {len(matching)} places in {city}:"]
    for p in matching:
        lines.append(f"- {p['name']} ({p['place_id']}): Type: {p['type']}, Rating: {p['rating']}/5")
    return "\n".join(lines)

@tool
def get_weather(city: str, days: int = 3) -> str:
    """Get the upcoming daily weather forecast for a city for a given number of days (1 to 7)."""
    coords = CITY_COORDINATES.get(city.lower())
    if not coords:
        return f"Sorry, I don't have coordinates for {city}."
    lat, lon = coords
    days = max(1, min(days, 7))
    url = (
        f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={lat}&longitude={lon}"
        f"&daily=temperature_2m_max,weathercode&timezone=auto"
    )
    try:
        resp = requests.get(url, timeout=10)
        resp.raise_for_status()
        data = resp.json()["daily"]
        lines = [f"Weather forecast for {city.title()} (next {days} days):"]
        for i in range(days):
            lines.append(
                f"- Day {i + 1} ({data['time'][i]}): {describe_weather_code(data['weathercode'][i])}, Max {data['temperature_2m_max'][i]}°C"
            )
        return "\n".join(lines)
    except Exception as e:
        return f"Could not fetch weather for {city}: {e}"

@tool
def estimate_budget(
    flight_price: int,
    hotel_price_per_night: int,
    nights: int,
    days: int,
    daily_local_expense: int = 1000,
) -> str:
    """Calculate the total estimated trip budget by summing flight cost, hotel stay, and daily food/local travel expenses."""
    hotel_total = hotel_price_per_night * nights
    local_total = daily_local_expense * days
    total_cost = flight_price + hotel_total + local_total
    return (
        f"Estimated Budget Breakdown:\n"
        f"- Flight: ₹{flight_price:,}\n"
        f"- Hotel ({nights} nights @ ₹{hotel_price_per_night:,}/night): ₹{hotel_total:,}\n"
        f"- Food & Local Travel ({days} days @ ₹{daily_local_expense:,}/day): ₹{local_total:,}\n"
        f"- Total Estimated Cost: ₹{total_cost:,}"
    )

st.set_page_config(page_title="AI Travel Planning Assistant", page_icon="✈️", layout="centered")
st.title("✈️ Agentic AI Travel Planning Assistant")
st.write("Plan a trip using real flight, hotel, and attraction datasets, live weather, and AI reasoning.")

api_key = os.environ.get("GOOGLE_API_KEY", "") or st.text_input(
    "Enter your Gemini API Key", type="password"
)

cities = ["Bangalore", "Delhi", "Mumbai", "Goa", "Chennai", "Hyderabad", "Kolkata", "Jaipur"]
col1, col2 = st.columns(2)
with col1:
    source = st.selectbox("From City", cities, index=0)
with col2:
    destination = st.selectbox("To City", cities, index=3)

days = st.slider("Trip Duration (Days)", min_value=1, max_value=7, value=3)
min_stars = st.selectbox("Minimum Hotel Stars", [1, 2, 3, 4, 5], index=4)
extra_notes = st.text_input("Preferences (optional)", value="I want the cheapest flight.")

if st.button("Generate Travel Plan"):
    if not api_key:
        st.error("Please provide a valid Gemini API key first.")
    else:
        os.environ["GOOGLE_API_KEY"] = api_key
        llm = ChatGoogleGenerativeAI(
            model="gemini-3.1-flash-lite",
            temperature=0,
            timeout=30,
            max_retries=1,
        )
        agent = create_agent(
            model=llm,
            tools=[search_flights, recommend_hotels, discover_places, get_weather, estimate_budget],
            system_prompt=(
                "You are an expert AI Travel Planning Assistant. Always call your tools "
                "for flights, hotels, places, weather, and budget before answering. "
                "Include sections: 1. Trip Summary, 2. Flight Option Selected, "
                "3. Hotel Recommendation, 4. Weather for Each Day, 5. Day-wise Itinerary, "
                "6. Estimated Budget Breakdown, and 7. Structured JSON Summary."
            ),
        )
        prompt = (
            f"Plan a {days}-day trip from {source} to {destination} with at least "
            f"{min_stars}-star hotel. {extra_notes}"
        )
        with st.spinner("Planning your trip..."):
            result = agent.invoke({"messages": [{"role": "user", "content": prompt}]})
        st.markdown(result["messages"][-1].text)
```

```python
%%writefile app.py
import json
import os
import time
from datetime import datetime
import requests
import streamlit as st
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from google.genai.errors import ServerError

# --- 1. Load Datasets ---
with open("flights.json", "r") as f:
    flights = json.load(f)

with open("hotels.json", "r") as f:
    hotels = json.load(f)

with open("places.json", "r") as f:
    places = json.load(f)

CITY_COORDINATES = {
    "delhi": (28.6139, 77.2090),
    "mumbai": (19.0760, 72.8777),
    "goa": (15.2993, 74.1240),
    "bangalore": (12.9716, 77.5946),
    "chennai": (13.0827, 80.2707),
    "hyderabad": (17.3850, 78.4867),
    "kolkata": (22.5726, 88.3639),
    "jaipur": (26.9124, 75.7873),
}

def describe_weather_code(code: int) -> str:
    """Convert Open-Meteo weather codes into readable labels."""
    if code == 0:
        return "Clear / Sunny"
    elif code in (1, 2, 3):
        return "Partly Cloudy"
    elif code in (45, 48):
        return "Foggy"
    elif code in (51, 53, 55, 61, 63, 65):
        return "Rainy"
    return "Showers / Unsettled"

# --- 2. Define Agent Tools ---
@tool
def search_flights(source: str, destination: str) -> str:
    """Search for flights between a source city and a destination city, and return the cheapest and fastest options."""
    matching = [
        f for f in flights
        if f["from"].lower() == source.lower() and f["to"].lower() == destination.lower()
    ]
    if not matching:
        return f"No flights found from {source} to {destination}."

    cheapest = min(matching, key=lambda f: f["price"])

    def duration_hrs(flight):
        dep = datetime.fromisoformat(flight["departure_time"])
        arr = datetime.fromisoformat(flight["arrival_time"])
        return (arr - dep).total_seconds() / 3600

    fastest = min(matching, key=duration_hrs)
    return (
        f"Found {len(matching)} flights from {source} to {destination}.\n"
        f"Cheapest: {cheapest['airline']} ({cheapest['flight_id']}) - ₹{cheapest['price']} "
        f"(Departs {cheapest['departure_time']}, Duration: {duration_hrs(cheapest):.1f} hrs)\n"
        f"Fastest: {fastest['airline']} ({fastest['flight_id']}) - ₹{fastest['price']} "
        f"(Departs {fastest['departure_time']}, Duration: {duration_hrs(fastest):.1f} hrs)"
    )

@tool
def recommend_hotels(city: str, min_stars: int = 1, max_price: int = 10000) -> str:
    """Find hotels in a given city filtered by minimum star rating and maximum price per night."""
    matching = [
        h for h in hotels
        if h["city"].lower() == city.lower()
        and h["stars"] >= min_stars
        and h["price_per_night"] <= max_price
    ]
    if not matching:
        return f"No hotels found in {city} with at least {min_stars} stars under ₹{max_price}/night."

    matching.sort(key=lambda h: (-h["stars"], h["price_per_night"]))
    lines = [f"Found {len(matching)} hotels in {city}:"]
    for h in matching:
        lines.append(
            f"- {h['name']} ({h['hotel_id']}): {h['stars']}-star, ₹{h['price_per_night']}/night | Amenities: {', '.join(h['amenities'])}"
        )
    return "\n".join(lines)

@tool
def discover_places(city: str, place_type: str = "", min_rating: float = 0.0) -> str:
    """Find tourist attractions in a city, optionally filtered by type and minimum rating."""
    matching = [
        p for p in places
        if p["city"].lower() == city.lower() and p["rating"] >= min_rating
    ]
    if place_type:
        matching = [p for p in matching if p["type"].lower() == place_type.lower()]
    if not matching:
        return f"No places found in {city} matching your criteria."

    matching.sort(key=lambda p: p["rating"], reverse=True)
    lines = [f"Found {len(matching)} places in {city}:"]
    for p in matching:
        lines.append(f"- {p['name']} ({p['place_id']}): Type: {p['type']}, Rating: {p['rating']}/5")
    return "\n".join(lines)

@tool
def get_weather(city: str, days: int = 3) -> str:
    """Get the upcoming daily weather forecast for a city for a given number of days (1 to 7)."""
    coords = CITY_COORDINATES.get(city.lower())
    if not coords:
        return f"Sorry, I don't have coordinates for {city}."
    lat, lon = coords
    days = max(1, min(days, 7))
    url = (
        f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={lat}&longitude={lon}"
        f"&daily=temperature_2m_max,weathercode&timezone=auto"
    )
    try:
        resp = requests.get(url, timeout=10)
        resp.raise_for_status()
        data = resp.json()["daily"]
        lines = [f"Weather forecast for {city.title()} (next {days} days):"]
        for i in range(days):
            lines.append(
                f"- Day {i + 1} ({data['time'][i]}): {describe_weather_code(data['weathercode'][i])}, Max {data['temperature_2m_max'][i]}°C"
            )
        return "\n".join(lines)
    except Exception as e:
        return f"Could not fetch weather for {city}: {e}"

@tool
def estimate_budget(
    flight_price: int,
    hotel_price_per_night: int,
    nights: int,
    days: int,
    daily_local_expense: int = 1000,
) -> str:
    """Calculate the total estimated trip budget by summing flight cost, hotel stay, and daily food/local travel expenses."""
    hotel_total = hotel_price_per_night * nights
    local_total = daily_local_expense * days
    total_cost = flight_price + hotel_total + local_total
    return (
        f"Estimated Budget Breakdown:\n"
        f"- Flight: ₹{flight_price:,}\n"
        f"- Hotel ({nights} nights @ ₹{hotel_price_per_night:,}/night): ₹{hotel_total:,}\n"
        f"- Food & Local Travel ({days} days @ ₹{daily_local_expense:,}/day): ₹{local_total:,}\n"
        f"- Total Estimated Cost: ₹{total_cost:,}"
    )

def invoke_agent_with_retry(
    primary_model: str, fallback_model: str, tools, system_prompt: str, prompt: str
):
  last_error = None
  for model_name in [primary_model, fallback_model]:
    llm = ChatGoogleGenerativeAI(
        model=model_name,
        temperature=0,
        timeout=30,
        max_retries=1,
    )
    agent = create_agent(model=llm, tools=tools, system_prompt=system_prompt)
    for attempt in range(1, 3):
      try:
        return agent.invoke({"messages": [{"role": "user", "content": prompt}]})
      except Exception as exc:
        last_error = exc
        msg = str(exc)
        is_transient = (
            "503" in msg
            or "504" in msg
            or "UNAVAILABLE" in msg
            or "DEADLINE_EXCEEDED" in msg
            or isinstance(exc, ServerError)
        )
        if is_transient and attempt < 2:
          time.sleep(2)
          continue
        if is_transient:
          break  # Try the fallback model
        raise
  raise last_error

# --- 3. Streamlit UI ---
st.set_page_config(page_title="AI Travel Planning Assistant", page_icon="✈️", layout="centered")
st.title("✈️ Agentic AI Travel Planning Assistant")
st.write("Plan a trip using real flight, hotel, and attraction datasets, live weather, and AI reasoning.")

api_key = os.environ.get("GOOGLE_API_KEY", "") or st.text_input(
    "Enter your Gemini API Key", type="password"
)

cities = ["Bangalore", "Delhi", "Mumbai", "Goa", "Chennai", "Hyderabad", "Kolkata", "Jaipur"]
col1, col2 = st.columns(2)
with col1:
    source = st.selectbox("From City", cities, index=0)
with col2:
    destination = st.selectbox("To City", cities, index=3)

days = st.slider("Trip Duration (Days)", min_value=1, max_value=7, value=3)
min_stars = st.selectbox("Minimum Hotel Stars", [1, 2, 3, 4, 5], index=4)
extra_notes = st.text_input("Preferences (optional)", value="I want the cheapest flight.")

if st.button("Generate Travel Plan"):
try:
  with st.spinner("Planning your trip..."):
    result = invoke_agent_with_retry(
        primary_model="gemini-3.1-flash-lite",
        fallback_model="gemini-2.5-flash-lite",
        tools=[
            search_flights,
            recommend_hotels,
            discover_places,
            get_weather,
            estimate_budget,
        ],
        system_prompt=(
            "You are an expert AI Travel Planning Assistant. Call the relevant tools "
            "once for flights, hotels, places, weather, and budget, then write a concise, "
            "complete response with sections 1 through 6 and section 7 ('7. Structured JSON Summary')."
        ),
        prompt=(
            f"Plan a {days}-day trip from {source} to {destination} with at least "
            f"{min_stars}-star hotel. {extra_notes}"
        ),
    )
  st.markdown(result["messages"][-1].text)
except Exception as e:
  st.error(f"Could not complete the request right now: {e}")
```

```python
%%writefile app.py
import json
import os
import time
from datetime import datetime
import requests
import streamlit as st
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from google.genai.errors import ServerError

# --- 1. Load Datasets ---
with open("flights.json", "r") as f:
    flights = json.load(f)

with open("hotels.json", "r") as f:
    hotels = json.load(f)

with open("places.json", "r") as f:
    places = json.load(f)

CITY_COORDINATES = {
    "delhi": (28.6139, 77.2090),
    "mumbai": (19.0760, 72.8777),
    "goa": (15.2993, 74.1240),
    "bangalore": (12.9716, 77.5946),
    "chennai": (13.0827, 80.2707),
    "hyderabad": (17.3850, 78.4867),
    "kolkata": (22.5726, 88.3639),
    "jaipur": (26.9124, 75.7873),
}

def describe_weather_code(code: int) -> str:
    """Convert Open-Meteo weather codes into readable labels."""
    if code == 0:
        return "Clear / Sunny"
    elif code in (1, 2, 3):
        return "Partly Cloudy"
    elif code in (45, 48):
        return "Foggy"
    elif code in (51, 53, 55, 61, 63, 65):
        return "Rainy"
    return "Showers / Unsettled"

# --- 2. Define Agent Tools ---
@tool
def search_flights(source: str, destination: str) -> str:
    """Search for flights between a source city and a destination city, and return the cheapest and fastest options."""
    matching = [
        f for f in flights
        if f["from"].lower() == source.lower() and f["to"].lower() == destination.lower()
    ]
    if not matching:
        return f"No flights found from {source} to {destination}."

    cheapest = min(matching, key=lambda f: f["price"])

    def duration_hrs(flight):
        dep = datetime.fromisoformat(flight["departure_time"])
        arr = datetime.fromisoformat(flight["arrival_time"])
        return (arr - dep).total_seconds() / 3600

    fastest = min(matching, key=duration_hrs)
    return (
        f"Found {len(matching)} flights from {source} to {destination}.\n"
        f"Cheapest: {cheapest['airline']} ({cheapest['flight_id']}) - ₹{cheapest['price']} "
        f"(Departs {cheapest['departure_time']}, Duration: {duration_hrs(cheapest):.1f} hrs)\n"
        f"Fastest: {fastest['airline']} ({fastest['flight_id']}) - ₹{fastest['price']} "
        f"(Departs {fastest['departure_time']}, Duration: {duration_hrs(fastest):.1f} hrs)"
    )

@tool
def recommend_hotels(city: str, min_stars: int = 1, max_price: int = 10000) -> str:
    """Find hotels in a given city filtered by minimum star rating and maximum price per night."""
    matching = [
        h for h in hotels
        if h["city"].lower() == city.lower()
        and h["stars"] >= min_stars
        and h["price_per_night"] <= max_price
    ]
    if not matching:
        return f"No hotels found in {city} with at least {min_stars} stars under ₹{max_price}/night."

    matching.sort(key=lambda h: (-h["stars"], h["price_per_night"]))
    lines = [f"Found {len(matching)} hotels in {city}:"]
    for h in matching:
        lines.append(
            f"- {h['name']} ({h['hotel_id']}): {h['stars']}-star, ₹{h['price_per_night']}/night | Amenities: {', '.join(h['amenities'])}"
        )
    return "\n".join(lines)

@tool
def discover_places(city: str, place_type: str = "", min_rating: float = 0.0) -> str:
    """Find tourist attractions in a city, optionally filtered by type and minimum rating."""
    matching = [
        p for p in places
        if p["city"].lower() == city.lower() and p["rating"] >= min_rating
    ]
    if place_type:
        matching = [p for p in matching if p["type"].lower() == place_type.lower()]
    if not matching:
        return f"No places found in {city} matching your criteria."

    matching.sort(key=lambda p: p["rating"], reverse=True)
    lines = [f"Found {len(matching)} places in {city}:"]
    for p in matching:
        lines.append(f"- {p['name']} ({p['place_id']}): Type: {p['type']}, Rating: {p['rating']}/5")
    return "\n".join(lines)

@tool
def get_weather(city: str, days: int = 3) -> str:
    """Get the upcoming daily weather forecast for a city for a given number of days (1 to 7)."""
    coords = CITY_COORDINATES.get(city.lower())
    if not coords:
        return f"Sorry, I don't have coordinates for {city}."
    lat, lon = coords
    days = max(1, min(days, 7))
    url = (
        f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={lat}&longitude={lon}"
        f"&daily=temperature_2m_max,weathercode&timezone=auto"
    )
    try:
        resp = requests.get(url, timeout=10)
        resp.raise_for_status()
        data = resp.json()["daily"]
        lines = [f"Weather forecast for {city.title()} (next {days} days):"]
        for i in range(days):
            lines.append(
                f"- Day {i + 1} ({data['time'][i]}): {describe_weather_code(data['weathercode'][i])}, Max {data['temperature_2m_max'][i]}°C"
            )
        return "\n".join(lines)
    except Exception as e:
        return f"Could not fetch weather for {city}: {e}"

@tool
def estimate_budget(
    flight_price: int,
    hotel_price_per_night: int,
    nights: int,
    days: int,
    daily_local_expense: int = 1000,
) -> str:
    """Calculate the total estimated trip budget by summing flight cost, hotel stay, and daily food/local travel expenses."""
    hotel_total = hotel_price_per_night * nights
    local_total = daily_local_expense * days
    total_cost = flight_price + hotel_total + local_total
    return (
        f"Estimated Budget Breakdown:\n"
        f"- Flight: ₹{flight_price:,}\n"
        f"- Hotel ({nights} nights @ ₹{hotel_price_per_night:,}/night): ₹{hotel_total:,}\n"
        f"- Food & Local Travel ({days} days @ ₹{daily_local_expense:,}/day): ₹{local_total:,}\n"
        f"- Total Estimated Cost: ₹{total_cost:,}"
    )

def invoke_agent_with_retry(primary_model: str, fallback_model: str, tools, system_prompt: str, prompt: str):
    last_error = None
    for model_name in [primary_model, fallback_model]:
        llm = ChatGoogleGenerativeAI(
            model=model_name,
            temperature=0,
            timeout=30,
            max_retries=1,
        )
        agent = create_agent(model=llm, tools=tools, system_prompt=system_prompt)
        for attempt in range(1, 3):
            try:
                return agent.invoke({"messages": [{"role": "user", "content": prompt}]})
            except Exception as exc:
                last_error = exc
                msg = str(exc)
                is_transient = (
                    "503" in msg
                    or "504" in msg
                    or "UNAVAILABLE" in msg
                    or "DEADLINE_EXCEEDED" in msg
                    or isinstance(exc, ServerError)
                )
                if is_transient and attempt < 2:
                    time.sleep(2)
                    continue
                if is_transient:
                    break
                raise
    raise last_error

# --- 3. Streamlit UI ---
st.set_page_config(page_title="AI Travel Planning Assistant", page_icon="✈️", layout="centered")
st.title("✈️ Agentic AI Travel Planning Assistant")
st.write("Plan a trip using real flight, hotel, and attraction datasets, live weather, and AI reasoning.")

api_key = os.environ.get("GOOGLE_API_KEY", "") or st.text_input(
    "Enter your Gemini API Key", type="password"
)

cities = ["Bangalore", "Delhi", "Mumbai", "Goa", "Chennai", "Hyderabad", "Kolkata", "Jaipur"]
col1, col2 = st.columns(2)
with col1:
    source = st.selectbox("From City", cities, index=0)
with col2:
    destination = st.selectbox("To City", cities, index=3)

days = st.slider("Trip Duration (Days)", min_value=1, max_value=7, value=3)
min_stars = st.selectbox("Minimum Hotel Stars", [1, 2, 3, 4, 5], index=4)
extra_notes = st.text_input("Preferences (optional)", value="I want the cheapest flight.")

if st.button("Generate Travel Plan"):
    if not api_key:
        st.error("Please provide a valid Gemini API key first.")
    else:
        os.environ["GOOGLE_API_KEY"] = api_key
        prompt = (
            f"Plan a {days}-day trip from {source} to {destination} with at least "
            f"{min_stars}-star hotel. {extra_notes}"
        )
        try:
            with st.spinner("Planning your trip..."):
                result = invoke_agent_with_retry(
                    primary_model="gemini-3.1-flash-lite",
                    fallback_model="gemini-2.5-flash-lite",
                    tools=[search_flights, recommend_hotels, discover_places, get_weather, estimate_budget],
                    system_prompt=(
                        "You are an expert AI Travel Planning Assistant. Call the relevant tools "
                        "once for flights, hotels, places, weather, and budget, then write a concise, "
                        "complete response with sections 1 through 6 and section 7 ('7. Structured JSON Summary')."
                    ),
                    prompt=prompt,
                )
            st.markdown(result["messages"][-1].text)
        except Exception as e:
            st.error(f"Could not complete the request right now: {e}")
```
